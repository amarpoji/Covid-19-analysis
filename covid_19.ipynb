In this project, I explore global COVID-19 statistics using a public dataset. The goal is to understand which countries were most affected, how cases changed over time, and how death and recovery rates vary by region.


```python
# importing all libraries related to project
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# load data set 
file_path = ('country_wise_latest.csv')
covid_19_dataset = pd.read_csv(file_path)

covid_19_dataset.head()
```

# Function for Barplot and Scatter Plot

```python


def plot_top_n_bar(data, column, n=10, title='', xlabel='', palette='viridis'):
    top_n = data.sort_values(column, ascending=False).head(n)
    plt.figure(figsize=(10,6))
    sns.barplot(data=top_n, x=column, y='Country/Region', palette=palette)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel('Country')
    plt.show()
    
def plot_scatter(data, x_col, y_col, hue_col, title):
    plt.figure(figsize=(10,6))
    sns.scatterplot(data=data, x=x_col, y=y_col, hue=hue_col, s=100, alpha=0.7)
    plt.title(title)
    plt.xlabel(x_col)
    plt.ylabel(y_col)
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.show()





  
```

### 📉 Recovery vs Death Rate (by WHO Region)

This scatter plot compares each country’s recovery rate and death rate. It is color-coded by WHO region to spot patterns across continents.

We can see that most countries fall into a predictable range, but outliers — with either extremely high or low recovery/death rates — may warrant special attention for unique conditions or reporting anomalies.


```python

plot_scatter(
    data=covid_19_dataset,
    x_col='Recovered / 100 Cases',
    y_col='Deaths / 100 Cases',
    hue_col='WHO Region',
    title='Recovery vs Death Rate by WHO Region'
)


```
